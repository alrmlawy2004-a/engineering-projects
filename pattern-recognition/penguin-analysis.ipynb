```python
import numpy as np
import pandas as pd
from palmerpenguins import load_penguins
```

```python
#Download Palmer Penguins data
df=load_penguins()
#Display a set of data
df.head(10)
```

```python
#To find out where the missing values ​​are located, count them in each column
df.isnull().sum()
```

```python
#Rows containing incomplete data were deleted.
df=df.dropna()
```

```python
#I converted the text columns to the category type.
df['species'] = df['species'].astype('category')
df['sex'] = df['sex'].astype('category')
df['island'] = df['island'].astype('category')
```

```python
#Remove irrelevant columns: The year column is not useful for analysis.
df = df.drop(columns=['year'])
```

```python
#Display data in general
df.describe()
```

```python
#Draw the distribution of all columns numerically
df.hist(figsize=(10,8))
```

These graphs show the distribution of measurements for four penguin characteristics (bill length, beak depth, flipper length, and body mass):

Bill length (mm): Most beaks range from 40 to 50 mm.

Bill depth (mm): Most beaks range from 16 to 19 mm.

Flipper length (mm): There are two distinct groups: a group with short flippers (around 190 mm) and a group with long flippers (around 215 mm). (This is the most important point.)

```python
import seaborn as sns
#Knowing how different the species are: If the species seem to have separate characteristics, it is good to distinguish between them.
sns.pairplot(df, hue='species')
```

The three types are clearly distinguished when using certain characteristics, especially when comparing:

Comparing body mass (body_mass_g) between species:

In simple terms: The diagram shows that Gentoo birds are the largest and heaviest of the three species, while Adelie and Chinstrap are very similar in weight.

Hypotheses:

H0: There is no difference in average fin length between species.

H1: There is a difference between at least one species.

Step 1: Data segmentation

Step 2: Calculating Averages

Step 3: Perform the ANOVA test

```python
adelie = df[df['species'] == 'Adelie']['flipper_length_mm']
gentoo = df[df['species'] == 'Gentoo']['flipper_length_mm']
chinstrap = df[df['species'] == 'Chinstrap']['flipper_length_mm']
```

```python
adelie.mean(), gentoo.mean(), chinstrap.mean()
```

```python
from scipy.stats import f_oneway

f_stat, p_value = f_oneway(adelie, gentoo, chinstrap)
f_stat, p_value
```

```python
#Create weight classes
df['mass_category'] = pd.cut(
    df['body_mass_g'],
    bins=[0, 3500, 4500, 6000],
    labels=['Light', 'Medium', 'Heavy'])
```

```python
#Create a cross-sectional table
table = pd.crosstab(df['species'], df['mass_category'])
```

```python
from scipy.stats import chi2_contingency

chi2, p, dof, expected = chi2_contingency(table)
chi2, p
```

p < 0.05: There is a relationship between gender and body weight.

p > 0.05: There is no relationship.

Ultimately, this study revealed clear biological differences between penguin species. Exploratory analysis and statistical testing confirmed that flipper length and body mass vary considerably among the species.
