```python
import matplotlib.pyplot as plt
import numpy as np

raw_data_gb = 10000       
replication_factor = 3   
compression_ratio_test = 4
network_speed_mbps = 100   
aws_s3_rate = 0.023      

compressed_size = raw_data_gb / compression_ratio_test
final_storage_size = compressed_size * replication_factor

total_bits = final_storage_size * 8 * 1024
replication_time_seconds = total_bits / network_speed_mbps
replication_time_hours = replication_time_seconds / 3600
monthly_bill = final_storage_size * aws_s3_rate
yearly_bill = monthly_bill * 12

print(f"Data analysis results (10 TB) ")
print(f"Final size after pressing and repetition: {final_storage_size:,.2f} GB")
print(f"Expected copy time via network: {replication_time_hours:.2f} hour")
print(f"Expected monthly bill: ${monthly_bill:.2f}")
print(f"Expected annual bill: ${yearly_bill:.2f}")


compression_ratios = np.arange(1, 6, 1)
storage_needed = (raw_data_gb / compression_ratios) * replication_factor
yearly_costs = storage_needed * aws_s3_rate * 12

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

ax1.plot(compression_ratios, storage_needed, marker='o', color='blue', linestyle='--')
ax1.set_title('Storage vs. Compression Ratio')
ax1.set_xlabel('Compression Ratio (x)')
ax1.set_ylabel('Total Storage (GB)')
ax1.grid(True)

ax2.plot(compression_ratios, yearly_costs, marker='s', color='red', linestyle='-')
ax2.set_title('Yearly Cost vs. Compression Ratio')
ax2.set_xlabel('Compression Ratio (x)')
ax2.set_ylabel('Yearly Cost ($)')
ax2.grid(True)

plt.tight_layout()
plt.show()
```
